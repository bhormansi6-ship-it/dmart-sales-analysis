```python

import re

class RailwayReservationSystem:

    def __init__(self):
        self.train_name = "Rajdhani Express"
        self.available_seats = 5
        self.passengers = []

    # Allocate seat
    def allocate_seat(self):
        if self.available_seats > 0:
            seat_number = 6 - self.available_seats
            seat = f"B{seat_number}"
            self.available_seats -= 1
            return seat
        else:
            return "No seats available"

    # Validate ticket details
    def validate_ticket(self, pnr, seat=None):
        pnr_pattern = r'^\d{8}$'
        seat_pattern = r'^[A-Z][0-9]{1,2}$'

        pnr_valid = re.fullmatch(pnr_pattern, str(pnr))
        seat_valid = (
            seat is None
            or seat == "None"
            or re.fullmatch(seat_pattern, str(seat))
        )

        return bool(pnr_valid and seat_valid)

    # Book ticket
    def book_ticket(self, name, pnr, seat=None, fare=0):

        if not self.validate_ticket(pnr, seat):
            print("Invalid PNR or seat number!")
            return

        if seat is None or seat == "None":
            seat = self.allocate_seat()

        if seat == "No seats available":
            print("No seats available for", name)
            return

        passenger = {
            "name": name,
            "pnr": str(pnr),
            "seat": seat,
            "fare": fare
        }

        self.passengers.append(passenger)
        print("Ticket booked successfully for", name)

    # Display passengers
    def display_passengers(self):
        print("\n--- PASSENGER DETAILS ---")

        for passenger in self.passengers:
            print(
                passenger["name"],
                "-",
                passenger["pnr"],
                "-",
                passenger["seat"],
                "- Rs.",
                passenger["fare"]
            )


# Create object
system = RailwayReservationSystem()

# Book tickets
system.book_ticket("Rahul Sharma", "43821756", None, 1500)
system.book_ticket("Priya Singh", "67934521", None, 1200)
system.book_ticket("Arjun Nair", "81204763", None, 1500)
system.book_ticket("Sunita Rao", "52983104", None, 1200)

# Display passenger details
system.display_passengers()

print("\nAvailable Seats:", system.available_seats)
```
