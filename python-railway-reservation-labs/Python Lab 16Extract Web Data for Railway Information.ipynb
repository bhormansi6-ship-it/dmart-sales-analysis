```python

import requests
from bs4 import BeautifulSoup


class RailwayScraper:

    def __init__(self):
        self.base_url = "https://contentitvedant.github.io/Railway_Booking/"

    def get_soup(self, page):
        url = self.base_url + page
        response = requests.get(url, timeout=15)
        response.raise_for_status()

        print("Fetched :", url, "| Status :", response.status_code)

        return BeautifulSoup(response.text, "html.parser")

    def get_train_schedule(self):
        soup = self.get_soup("trains.html")
        table = soup.find(id="train-schedule-table")
        trains = []

        for row in table.find("tbody").find_all("tr"):
            cells = row.find_all("td")

            trains.append({
                "train_no": cells[0].get_text(strip=True),
                "train_name": cells[1].get_text(strip=True),
                "from": cells[2].get_text(strip=True),
                "to": cells[3].get_text(strip=True),
                "departure": cells[4].get_text(strip=True),
                "arrival": cells[5].get_text(strip=True),
                "duration": cells[6].get_text(strip=True)
            })

        return trains

    def get_passengers(self):
        soup = self.get_soup("passengers.html")
        table = soup.find(id="passenger-records-table")
        passengers = []

        for row in table.find("tbody").find_all("tr"):
            cells = row.find_all("td")

            passengers.append({
                "pnr": cells[0].get_text(strip=True),
                "name": cells[1].get_text(strip=True),
                "date": cells[2].get_text(strip=True),
                "class": cells[3].get_text(strip=True),
                "seat": cells[4].get_text(strip=True),
                "fare": cells[5].get_text(strip=True)
            })

        return passengers

    def get_seat_availability(self):
        soup = self.get_soup("seat-availability.html")
        table = soup.find(id="seat-availability-table")
        seats = []

        for row in table.find("tbody").find_all("tr"):
            cells = row.find_all("td")

            seats.append({
                "train_no": cells[0].get_text(strip=True),
                "train_name": cells[1].get_text(strip=True),
                "seats_sl": cells[2].get_text(strip=True),
                "seats_3a": cells[3].get_text(strip=True),
                "seats_2a": cells[4].get_text(strip=True),
                "seats_1a": cells[5].get_text(strip=True)
            })

        return seats


scraper = RailwayScraper()

print("\n--- TRAIN SCHEDULE ---")
for train in scraper.get_train_schedule():
    print(
        train["train_no"], "|",
        train["train_name"], "|",
        train["from"], "->", train["to"], "|",
        train["duration"]
    )

print("\n--- PASSENGER RECORDS ---")
for passenger in scraper.get_passengers():
    print(
        "PNR:", passenger["pnr"], "|",
        "Name:", passenger["name"], "|",
        "Class:", passenger["class"], "|",
        "Seat:", passenger["seat"]
    )

print("\n--- SEAT AVAILABILITY ---")
for seat in scraper.get_seat_availability():
    print(
        "Train", seat["train_no"], "-",
        seat["train_name"], "|",
        "SL:", seat["seats_sl"], "|",
        "3A:", seat["seats_3a"], "|",
        "2A:", seat["seats_2a"]
    )
```

```python
trains = scraper.get_train_schedule()

with open("train_schedule.txt", "w", encoding="utf-8") as file:
    for train in trains:
        file.write(
            train["train_no"] + " | " +
            train["train_name"] + " | " +
            train["from"] + " -> " +
            train["to"] + " | " +
            train["duration"] + "\n"
        )

print("Train schedule saved successfully!")
```

```python

import csv

trains = scraper.get_train_schedule()

with open("train_schedule.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow([
        "train_no",
        "train_name",
        "from",
        "to",
        "departure",
        "arrival",
        "duration"
    ])

    for t in trains:
        writer.writerow([
            t["train_no"],
            t["train_name"],
            t["from"],
            t["to"],
            t["departure"],
            t["arrival"],
            t["duration"]
        ])

print("CSV file created successfully.")


     
```
